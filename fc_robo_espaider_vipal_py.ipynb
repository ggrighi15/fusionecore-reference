import pandas as pd # Import pandas to process the scraped data after export

# Configuração simulada de ambiente (Em produção, viria do .env)
# Certifique-se de que estas variáveis de ambiente estão configuradas corretamente
# para o seu ambiente Colab ou local.
os.environ["ESPAIDER_URL"] = "https://www.espaider.com.br/vipal"
os.environ["ESPAIDER_USER"] = "Gustavo_r" # Substitua pelo seu usuário real
os.environ["ESPAIDER_PASS"] = "Gusgic@790315" # Substitua pela sua senha real

class RoboEspaiderVipal:
    def __init__(self):
        self.options = Options()
        self.options.add_argument("--headless")  # Execução sem interface gráfica
        self.options.add_argument("--no-sandbox")
        self.options.add_argument("--disable-dev-shm-usage")
        self.options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) FusioneCoreBot/5.1")
        self.driver = None
        self.results = []

    def start(self):
        print(f"[{datetime.now()}] 🚀 Iniciando Driver Chrome...")
        self.driver = webdriver.Chrome(service=webdriver.ChromeService(ChromeDriverManager().install()), options=self.options)

    def login(self):
        url = os.getenv("ESPAIDER_URL")
        print(f"[{datetime.now()}] 🔗 Acessando {url}...")
        self.driver.get(url)

        # REALIZAR LOGIN:
        # Substitua 'username_field_id', 'password_field_id' e 'login_button_id'
        # pelos seletores (ID, NAME, XPATH, etc.) corretos da página de login da Vipal.
        # Use o inspetor de elementos do navegador para encontrá-los.
        print(f"[{datetime.now()}] 🔑 Autenticando usuário {os.getenv('ESPAIDER_USER')}...")
        try:
            # Espera até que o campo de usuário esteja presente e interage
            WebDriverWait(self.driver, 20).until(
                EC.presence_of_element_located((By.ID, "username_field_id")) # <-- SUBSTITUA Pelo ID do campo de usuário
            ).send_keys(os.getenv("ESPAIDER_USER"))

            # Espera até que o campo de senha esteja presente e interage
            self.driver.find_element(By.ID, "password_field_id").send_keys(os.getenv("ESPAIDER_PASS")) # <-- SUBSTITUA Pelo ID do campo de senha

            # Espera até que o botão de login esteja clicável e clica
            WebDriverWait(self.driver, 10).until(
                EC.element_to_be_clickable((By.ID, "login_button_id")) # <-- SUBSTITUA Pelo ID do botão de login
            ).click()

            # Aguarda a mudança de URL ou a presença de um elemento pós-login para confirmar
            # Ajuste o URL ou o seletor do elemento pós-login conforme necessário
            WebDriverWait(self.driver, 20).until(
                EC.url_changes(url) # Verifica se a URL mudou após o login (ex: para o dashboard)
            )
            print(f"[{datetime.now()}] ✅ Login realizado com sucesso.")
        except Exception as e:
            print(f"[{datetime.now()}] ❌ Erro durante o login. Verifique os seletores e suas credenciais. Erro: {e}")
            raise # Re-raise a exceção para interromper se o login falhar

    def scrape_processos(self):
        print(f"[{datetime.now()}] 📂 Navegando para módulo Contencioso...")
        try:
            # NAVEGAÇÃO PARA O MÓDULO CONTENCIOSO:
            # Substitua o XPath abaixo pelo seletor correto para o link ou botão
            # que leva ao módulo 'Contencioso'. Pode ser um ID, XPATH, LINK_TEXT, etc.
            contencioso_link = WebDriverWait(self.driver, 15).until(
                EC.element_to_be_clickable((By.XPATH, "//a[contains(text(), 'Contencioso')]")) # <-- SUBSTITUA Pelo seletor do link/botão "Contencioso"
            )
            contencioso_link.click()
            print(f"[{datetime.now()}] ✅ Módulo Contencioso acessado.")

            # EXTRAÇÃO DOS DADOS DA TABELA:
            # Substitua 'table_processos' pelo ID, XPATH, ou outro seletor da tabela de processos.
            table = WebDriverWait(self.driver, 20).until(
                EC.presence_of_element_located((By.ID, "table_processos")) # <-- SUBSTITUA Pelo ID da tabela de processos
            )

            # Encontra todas as linhas da tabela, geralmente pulando a primeira (cabeçalho)
            # Ajuste [1:] se o cabeçalho não for a primeira linha ou se não houver cabeçalho.
            rows = table.find_elements(By.TAG_NAME, "tr")[1:]

            extracted_data = []
            print(f"[{datetime.now()}] 🕷️ Extraindo dados da tabela principal...")

            for i, row in enumerate(rows):
                try:
                    # Encontra todas as células de dados (<td>) dentro de cada linha.
                    # A ordem das colunas é crucial e deve corresponder aos seus nomes.
                    # Exemplo: id_espaider, numero_cnj, parte_contraria, assunto, valor_causa, status, ultima_movimentacao
                    cols = row.find_elements(By.TAG_NAME, "td")
                    if len(cols) >= 7: # Verifica se há colunas suficientes para todos os dados esperados
                        id_espaider = cols[0].text.strip()
                        numero_cnj = cols[1].text.strip()
                        parte_contraria = cols[2].text.strip()
                        assunto = cols[3].text.strip()

                        # Tratamento para 'valor_causa': remove "R$", "." e substitui "," por "."
                        valor_causa_str = cols[4].text.strip().replace("R$", "").replace(".", "").replace(",", ".").strip()
                        valor_causa = float(valor_causa_str) if valor_causa_str else 0.0

                        status = cols[5].text.strip()
                        ultima_movimentacao = cols[6].text.strip()

                        extracted_data.append({
                            "id_espaider": id_espaider,
                            "numero_cnj": numero_cnj,
                            "parte_contraria": parte_contraria,
                            "assunto": assunto,
                            "valor_causa": valor_causa,
                            "status": status,
                            "ultima_movimentacao": ultima_movimentacao
                        })
                    else:
                        print(f"[{datetime.now()}] ⚠️ Linha {i+1} não tem colunas suficientes para extração completa, pulando.")

                except Exception as row_error:
                    print(f"[{datetime.now()}] ❌ Erro ao processar linha {i+1}: {row_error}")
                    continue # Continua para a próxima linha mesmo que uma falhe

            self.results = extracted_data
            print(f"[{datetime.now()}] 📦 {len(self.results)} processos extraídos e normalizados.")

        except Exception as e:
            print(f"[{datetime.now()}] ❌ Erro crítico durante a extração de processos do Contencioso. Erro: {e}")
            self.results = [] # Garante que os resultados estejam vazios em caso de falha crítica
            raise # Re-raise a exceção para indicar falha

    def export_data(self):
        filename = "export_espaider_vipal.json"
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump(self.results, f, indent=2, ensure_ascii=False)
        print(f"[{datetime.now()}] 💾 Dados salvos em {filename}")

    def stop(self):
        if self.driver:
            self.driver.quit()
        print(f"[{datetime.now()}] 🏁 Sessão encerrada.")

if __name__ == "__main__":
    bot = RoboEspaiderVipal()
    try:
        bot.start()
        bot.login()
        bot.scrape_processos()
        bot.export_data()

        # Carregar e exibir os dados extraídos para verificação
        df_real_data = pd.read_json("export_espaider_vipal.json")

        print(f"Total de processos extraídos: {len(df_real_data)}")

    except Exception as e:
        print(f"❌ Erro crítico na execução do bot: {e}")
    finally:
        bot.stop()
